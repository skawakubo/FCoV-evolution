```python
import pandas as pd

# 1. 読み込むファイル名リスト
files = {
    '2344': '2344_branch_mutation_summary.tsv',
    '3129': '3129_branch_mutation_summary.tsv',
    '3423': '3423_branch_mutation_summary.tsv',
    '3558': '3558_branch_mutation_summary.tsv',
    '4109': '4109_branch_mutation_summary.tsv'
}

# 2. リスト内包表記でデータを読み込み、ID列を追加
dfs = []
for nrr_id, file in files.items():
    df = pd.read_csv(file, sep='\t')
    df['nrr_id'] = nrr_id  # どのNRR由来かを示す列を追加
    dfs.append(df)

# 3. 結合（concat）
combined_df = pd.concat(dfs, ignore_index=True)

# 結果の表示
print(combined_df.head())

# 4. まとめたものを保存
combined_df.to_csv('combine_NRR_branch_mutation_summary.tsv', sep='\t', index=False)
```

変異した後にfecv_consensusと一致する配列に戻る変異を獲得したもの

```python
from Bio import SeqIO
import pandas as pd

# FASTAファイルのパス
fasta_path = '/Users/izumikida/FCoV_seq_relatives/FCoV1_S_final/snipit_NRR_number/fecv_consensus.fas'

# FASTAを読み込む
record = SeqIO.read(fasta_path, "fasta")

# 各ポジションとState（アミノ酸/塩基）を抽出
data = {
    'pos': range(1, len(record.seq) + 1),  # Pos: 1から始まる通し番号
    'state2': list(record.seq)            # State2: その位置の配列文字
}

# データフレーム作成
df_consensus = pd.DataFrame(data)
df_consensus
```

```python
import pandas as pd

# 1. コンセンサスデータのカラムを明確にリネームする
# コンセンサスの 'state2' が実はコンセンサス配列の文字であるため
df_consensus = df_consensus.rename(columns={'state2': 'consensus_state'})

# 2. マージ実行
# branch_mut と df_consensus を pos をキーに結合
merged_df = pd.merge(branch_mut, df_consensus, on='pos', how='inner')

# 3. state2 (変異後の文字) と consensus_state (コンセンサス配列の文字) を比較
match_df = merged_df[merged_df['state2'] == merged_df['consensus_state']]

# 結果の表示
print(match_df)
```
