```python
def fasta_to_dataframe(fasta_file):
    ids = []
    sequences = []
    for record in SeqIO.parse(fasta_file, "fasta"):
        ids.append(record.id)
        sequences.append(str(record.seq))

    df = pd.DataFrame({
        "ID": ids,
        "seq": sequences
    })
    return df
```

```python
# ORFのpositionを定義する関数 parse_positions
def parse_positions(pos_file):
    ORFs = {}
    with open(pos_file, 'r') as f:
        for line in f:
            line = line.strip()
            if not line:  # skip empty line
                continue
            parts = line.split(',')
            # parts[0] = ORF name, parts[1] = start pos, parts[2] = end pos
            orfname = parts[0]
            start = int(parts[1])
            end = int(parts[2])
            ORFs[orfname] = [start, end]
    return ORFs
FCoV_pos_path = '/Users/izumikida/FCoV_seq_relatives/FCoV1_extract_ORF/LC880185.1_pos.csv'
FCoV_pos = parse_positions(FCoV_pos_path)
FCoV_pos
```

ORF毎にaliginment後のpositionで切り分ける　ATGで始まりが揃っているかと終わりが終始コドンで揃っているか確認
posがズレるので、pos指定で1ab, 3c, 7bのみ切り出す。

```python
from Bio import SeqIO
from Bio.Seq import Seq

def extract_specific_ORFs_with_gaps(orfd, wg_al_file, translate='Y'):
    recdict = SeqIO.to_dict(SeqIO.parse(wg_al_file, 'fasta'))

    for orf, (start_pos, end_pos) in orfd.items():
        # インデックス調整 (1-based to 0-based)
        start = start_pos - 1
        end = end_pos
        
        output_file = f'extract_{orf}.fasta'
        aa_file = f'extract_{orf}_aa.fas'
        
        fasta_content = []
        aa_content = []
        
        for virname in recdict:
            # 配列の切り出し (Gapを維持する)
            thisseq = str(recdict[virname].seq)[start:end]
            fasta_content.append(f'>{virname}\n{thisseq}')
            
            if translate == 'Y':
                # 翻訳時はGapを無視してアミノ酸に変換
                aaseq = Seq(thisseq.replace('-', '')).translate()
                aa_content.append(f'>{virname}\n{str(aaseq)}')
        
        # NTファイル出力 (Gapあり)
        with open(output_file, 'w') as out:
            out.write('\n'.join(fasta_content))
        print(f"Saved (Gap kept): {output_file}")
            
        if translate == 'Y':
            with open(aa_file, 'w') as out:
                out.write('\n'.join(aa_content))
            print(f"Saved: {aa_file}")

# --- 設定と実行 ---
target_ORFs = {
    'ORF1ab': (387, 20536),
    '3c': (25418, 25615),
    '7b': (28694, 29315)
}

input_fasta = 'FCoV1_full_ORF_20260311.linsi.fasta'
extract_specific_ORFs_with_gaps(target_ORFs, input_fasta, translate='Y')
```

```python
import pandas as pd
from Bio import SeqIO

def get_accession_table(file_list):
    all_data = []
    
    for f in file_list:
        for rec in SeqIO.parse(f, 'fasta'):
            # ATGC- 以外の文字数をカウント
            ambig_count = sum(1 for base in str(rec.seq).upper() if base not in 'ATGC-')
            all_data.append({
                'File': f,
                'Accession': rec.id,
                'Ambiguous_Count': ambig_count
            })
            
    return pd.DataFrame(all_data)

# 実行
files = ['extract_3c_cds.fasta']
df_detail = get_accession_table(files)

# 混合塩基が含まれる配列のみに絞って表示（0のものは除外する場合）
df_dirty = df_detail[df_detail['Ambiguous_Count'] > 0]
print(df_dirty)
```
