```python
# merge data for lolipop plot

import pandas as pd

# 1. データの読み込み
df_main = pd.read_csv("volcano_input.csv")
df_m1058 = pd.read_csv("volcano_input_M1058.csv")
df_fcs = pd.read_csv("volcano_input_FCS.csv")

# 2. 優先順位に従ってリスト化する
# 最初に置いたdf（ここではdf_main）のデータが優先的に保持されます
dfs = [df_main, df_m1058, df_fcs]

# 3. 縦に連結 (concat)
df_merged = pd.concat(dfs, ignore_index=True)

# 4. 'pos' 列で重複を削除
# keep='first' にすることで、先に読み込まれた df_main のデータが残ります
df_final = df_merged.drop_duplicates(subset=['pos'], keep='first')

# 5. 結果の保存（必要に応じて）
df_final.to_csv("lplipop.csv", index=False)

print(f"Total merged records: {len(df_final)}")
```
