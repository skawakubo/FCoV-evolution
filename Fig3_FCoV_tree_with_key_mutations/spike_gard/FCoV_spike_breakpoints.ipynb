# Identifying recombination breakpoints for FCoV spike

Full analysis for mapping GARD breakpoints run based on clu90 alignment to final complete alignment `FCoV1_S_all_seq_cleaned_cds_linsi_curated_fullllength_fix_dates.fasta`

```python
from Bio import SeqIO
```

```python
def thelens(thel):
    for i in range(len(thel)-1):
        print(thel[i+1] - thel[i])
        
def rm_shorts(thel, thresh):
    newl = [thel[0]]
    for i in range(len(thel)-1):
        if thel[i+1] - thel[i] > thresh:
            newl.append(thel[i+1])
    return newl



def mapals_updated(posl, alone, altwo, refname):
    records1 = SeqIO.to_dict(SeqIO.parse(alone, 'fasta'))
    refseq1 = str(records1[refname].seq)
    
    rawpos = []
    for p in posl:
        seqseg = refseq1[:p]
        seqseg = seqseg.replace('-', '')
        rawpos.append(len(seqseg))
        
    records2 = SeqIO.to_dict(SeqIO.parse(altwo, 'fasta'))
    refseq2 = str(records2[refname].seq)
    
    themap = {}
    c = 0
    for i in range(len(refseq2)):
        if str(c) not in themap:
            themap[str(c)] = [i]
        else:
            themap[str(c)].append(i)
        if refseq2[i] != '-':
            c += 1
    
    finalpos = [max(themap[str(x)]) for x in rawpos]
    return finalpos
```
