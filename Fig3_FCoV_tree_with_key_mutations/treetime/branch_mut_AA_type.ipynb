変異した後にfecv_consensusと一致する配列に戻る変異を獲得したもの

```python
from Bio import SeqIO
import pandas as pd

# FASTAファイルのパス
fasta_path = '/Users/izumikida/FCoV-evolution/Fig2_FIPV_vs_FECV_Fishers_exact_test/Spike/fecv_consensus.fas'

# FASTAを読み込む
record = SeqIO.read(fasta_path, "fasta")

# 各ポジションとState（アミノ酸/塩基）を抽出
data = {
    'pos': range(1, len(record.seq) + 1),  # Pos: 1から始まる通し番号
    'state2': list(record.seq)            # State2: その位置の配列文字
}

# データフレーム作成
df_consensus = pd.DataFrame(data)
df_consensus
```

```python
import pandas as pd

# 1. コンセンサスデータのカラムを明確にリネームする
# コンセンサスの 'state2' が実はコンセンサス配列の文字であるため
df_consensus = df_consensus.rename(columns={'state2': 'consensus_state'})

# 2. マージ実行
# branch_mut と df_consensus を pos をキーに結合
merged_df = pd.merge(branch_mut, df_consensus, on='pos', how='inner')

# 3. state2 (変異後の文字) と consensus_state (コンセンサス配列の文字) を比較
match_df = merged_df[merged_df['state2'] == merged_df['consensus_state']]

# 結果の表示
print(match_df)
```
