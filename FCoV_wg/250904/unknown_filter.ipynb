```python
from Bio import SeqIO

fasta_in = "/Users/izumikida/FCoV-evolution/FCoV_wg/250903/FCoV2_extractORF/al_FCoV2_comp_S_nt_prelim.fas"
drop_file = "/Users/izumikida/FCoV-evolution/FCoV_wg/250903/al_FCoV1_FCoV2_nt_prelim_modified/drop.list"
fasta_out = "/Users/izumikida/FCoV-evolution/FCoV_wg/250904/al_FCoV2_comp_S_nt_prelim.filtered.fas"

# drop.list のアクセッション番号を読み込み
with open(drop_file) as f:
    drop_ids = {line.strip() for line in f if line.strip()}

# フィルタリングして書き出し
with open(fasta_out, "w") as out:
    for record in SeqIO.parse(fasta_in, "fasta"):
        if record.id not in drop_ids:
            SeqIO.write(record, out, "fasta")

print(f"Filtered FASTA written to: {fasta_out}")
```
